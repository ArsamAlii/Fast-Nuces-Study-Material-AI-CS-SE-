# Breath First Search

```python
import networkx as nx
from queue import Queue
import matplotlib.pyplot as plt

def bfs(graph, start, goal):
    visited = set()
    queue = Queue()
    queue.put((start, [start]))

    while not queue.empty():
        node, path = queue.get()
        if node == goal:
            return path
        visited.add(node)

        for neighbor in graph.neighbors(node):
            if neighbor not in visited:
                queue.put((neighbor, path + [neighbor]))

G = nx.Graph()
G.add_weighted_edges_from([(1, 2, 1.5), (2, 4, 2.0), (2, 5, 2.5), (1, 3, 1.0), (3, 6, 3.0), (6, 8, 1.2)])

pos = nx.spring_layout(G)
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw(G, pos, with_labels=True, node_color='lightblue', node_size=1000)
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

start_node = 1
goal_node = 6
path = bfs(G, start_node, goal_node)

if path:
    print(f"BFS path from node {start_node} to node {goal_node}: {path}")
else:
    print(f"No path found from node {start_node} to node {goal_node}.")

plt.show()
```

## Depth First Search

```python
import networkx as nx

def dfs(graph, start, goal):
    visited = set()
    stack = [(start, [start])]

    while stack:
        node, path = stack.pop()
        if node == goal:
            return path
        if node not in visited:
            visited.add(node)
            for neighbor in reversed(list(graph.neighbors(node))): 
                if neighbor not in visited:
                    stack.append((neighbor, path + [neighbor]))

    return None

G = nx.Graph()
G.add_edges_from([(1, 2), (2, 4), (2, 5), (1, 3), (3, 6), (6, 8)])

nx.draw(G, with_labels=True, node_color='lightblue', node_size=1000)

start_node = 1
goal_node = 6
path = dfs(G, start_node, goal_node)

if path:
    print(f"DFS path from node {start_node} to node {goal_node}: {path}")
else:
    print(f"No path found from node {start_node} to node {goal_node}.")
```
