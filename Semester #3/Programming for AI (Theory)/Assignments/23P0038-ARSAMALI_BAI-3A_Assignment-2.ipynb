`Q-3` Merge the current dataset with a second DataFrame containing Location and their respective Region (North, South, East, West). After merging, perform the following:
<ol>
<li>Filter the data to include only transactions made by customers aged "25-45" who availed a discount greater than ₹100 using a "Credit Card."</li>
<li>For each Region and Product Category, calculate the total Net Amount and average Discount Amount (INR).</li>
<li>Find the top 2 most frequently purchased Product Categories for each Region.</li>
<li>Sort the results by Region alphabetically.</li>
</ol>

````python
#The following error solved from stack over flow https://stackoverflow.com/questions/46227170/not-supported-between-instances-of-str-and-int:
# ['>=' not supported between instances of 'str' and 'int'] 

#df['Age Group'] = pd.to_numeric(df['Age Group'])-->did not worked
#--------------------------------------------------------------------------

df = pd.read_csv(r"D:\DESKTOP\project1_df.csv")
region_data = {
    'Location': ['Ahmedabad', 'Bangalore', 'Delhi', 'Mumbai', 'Chennai', 'Pune', 'Hyderabad', 'Kolkata', 'Jaipur', 'Lucknow'],
    'Region': ['West', 'South', 'North', 'West', 'South', 'West', 'South', 'East', 'North', 'North']
}

second_df = pd.DataFrame(region_data)

merged_df=pd.merge(df,second_df,on='Location',how='left')

#print(merged_df)
merged_df['Age Group'] = merged_df['Age Group'].astype(str).str.extract(r'(\d+)').astype(float)
filtered_df = merged_df[(merged_df['Age Group'].between(25, 45)) & 
                        (merged_df['Discount Amount (INR)'] > 100) & 
                        (merged_df['Purchase Method'] == 'Credit Card')]


print("FILTERED DATA FRAME\n\n",filtered_df)
#----------------------------------------------
total_net_amount=merged_df.groupby(['Region', 'Product Category'])['Net Amount'].sum().reset_index()
total_net_amount=total_net_amount.rename(columns={'Net Amount':'Total_Net_Amount'})

avg_discount_amount=merged_df.groupby(['Region','Product Category'])['Discount Amount (INR)'].mean().reset_index()
avg_discount_amount=avg_discount_amount.rename(columns={'Discount Amount (INR)':'Avg_Discount_Amount'})

grouped_summary=pd.merge(total_net_amount,avg_discount_amount,on=['Region','Product Category'])
print("\n\nTotal Net Amount and Average Discount Amount per Region and Product Category:\n\n", grouped_summary)
#----------------------------------------------
# Calculate frequency of each Product Category in each Region

frequency_df=filtered_df.groupby(['Region', 'Product Category']).size().reset_index(name='Frequency')
frequency_df=frequency_df.sort_values(['Region', 'Frequency'], ascending=[True,False])
top_categories=frequency_df.groupby('Region').head(2)
print("\n\nTop 2 Most Frequently Purchased Product Categories for Each Region:\n",top_categories)

#=------------------------------------------------------------
grouped_summary = grouped_summary.sort_values('Region')
top_categories = top_categories.sort_values('Region')

print("\n\nGrouped Summary:\n",grouped_summary)
print("\n\nTop Categories:\n",top_categories)
````

`Q-4` Use pd.cut to categorize the transactions into the following purchase year bins: <br>
<ul>
<li>2018 and earlier <strong>( Will have 0 Values )</strong></li>
<li>2019-2020 <strong>( 14181 Values )</strong></li>
<li>2021-2022<strong>( 21938 Values )</strong></li>
<li>2023-2024<strong>( 18881 Values )</strong></li>
</ul>
Visualize the frequency of transactions in each bin using a bar plot. For Now use pandas built in plotting method ( .plot( kind="bar" ) ). 

Hint: Take year from "Purchase Date".

````python
# method help to extract year is taken feom following link:
#https://www.geeksforgeeks.org/get-month-and-year-from-date-in-pandas-python/

#--------------------
import matplotlib.pyplot as plt
df['Purchase Date'] = pd.to_datetime(df['Purchase Date'],errors='coerce')
df['Year']=df['Purchase Date'].dt.year

bins=[0,2018,2020,2022,2024]

labels= ['>2018','2019-2020','2021-2022','2023-2024']
df['Year Bin']=pd.cut(df['Year'],bins=bins,labels=labels)

transaction_count_by_bin=df['Year Bin'].value_counts().sort_index()

transaction_count_by_bin.plot(kind='bar',color='blue')

plt.xlabel('Purchase Year Bins')
plt.ylabel('Number of Transactions')
plt.title('Frequency of Transactions by Year Bin')

plt.show()
````

`Q-5` Create a pivot table showing the mean Net Amount for each Gender and Product Category, including only transactions where the Net Amount is greater than ₹2000 and a discount was availed.

Next, use the melt function to reshape the pivot table into a long format, with each row representing a combination of Gender, Product Category, and the corresponding mean Net Amount.

````python
filtered2 = df[(df['Net Amount'] > 2000) & (df['Discount Amount (INR)'] > 0)]

pt= pd.pivot_table(filtered2, values='Net Amount', index='Gender', columns='Product Category', aggfunc='mean')

melted_df = pd.melt(pt.reset_index(), id_vars='Gender', var_name='Product Category', value_name='Mean Net Amount')
melted_df
````
