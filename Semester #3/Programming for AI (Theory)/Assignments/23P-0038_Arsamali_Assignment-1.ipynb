### <u>Question 1:</u>



You have a list of dictionaries, where each dictionary represents a transaction in a dataset. Write a program to group the transactions by `Store_Type` and find the total sales for each store.

**Input:**

```python
transactions = [
    {'Transaction_ID': 1, 'Store_Type': 'Electronics', 'Total_Sales': 950},
    {'Transaction_ID': 2, 'Store_Type': 'Books', 'Total_Sales': 100},
    # More transactions...
]
```
##### Note

The `transactions` list will change on every output due to `np.random`. In the future, you will explore `numpy`, so don't worry about it now.

Also, explore the Python `"random"` library if possible: [Python Random Module](https://www.geeksforgeeks.org/python-random-module/)


````python
import numpy as np

transactions=[]
for i in range(1, 11):
    data={"Transaction_ID": i}
    data["Store_Type"] = np.random.choice(["Electronics", "Books", "Stationary", "Clothing", "Grocery", "Furniture", "Pharmacy"])
    data["Total_Sales"] = np.random.choice(np.arange(100, 1001, 50))
    transactions.append(data)

print("Transactions:")
print(transactions)

def storetype(transactions):
    st_type={}

    for transaction in transactions:
        store_type=transaction['Store_Type']
        total_sale=transaction['Total_Sales']

        if store_type in st_type:
            st_type[store_type]+=total_sale
        else:
            st_type[store_type]=total_sale

    return st_type

grouped_sales=storetype(transactions)

print()
print("total sales in store by type:")
for store_type, total_sales in grouped_sales.items():
    print(f"{store_type}: ${total_sales}")
````

### <u>Question 5:</u>

You have a list of floats representing daily transaction amounts. Use reduce(),filter() and a lambda function to calculate the sum of all the transactions that are greater than $100.

```python
[110.46, 75.33, 120.43, 94.72]
```

```python

```

Explore Reduce function in python: [Python Reduce Function](https://www.geeksforgeeks.org/reduce-in-python/)

````python
from functools import reduce
transactions=[110.46, 75.33, 120.43, 94.72]


f_transactions=filter(lambda x:x>100,transactions)


total_sum=reduce(lambda x,y:x+y,f_transactions)

print(total_sum)
````

**input:**
```python

```
**Output:**

```python
List-2
Set-2
Tuples-1
Others-1

````python
lst=[{'Course', 'Lab'},{'Prog', 'AI'},('1', '2'),['c', 'd'],['10', '20'],"PAI"]

set_num=0
list_num=0
tuple_num=0
other=0

for i in range(0,len(lst)):
    if(type(lst[i])==set):
        set_num=set_num+1
    elif(type(lst[i])==list):
        list_num=list_num+1
    elif(type(lst[i])==tuple):  
        tuple_num=tuple_num+1
    else:
        other=other+1


print("number of sets are :",set_num)
print("number of lists are : ",list_num)
print("number of tuples are : ",tuple_num)
print("other:",other)


````

```

Explore *args and **kwargs in python : [Python *args and **kwargs](https://www.geeksforgeeks.org/args-kwargs-python/)

````python
def concatenate_dicts(*dicts):
    result=dict()

    for dic in dicts:
        for key, value in dic.items():
            if key in result:
                result[key]+=value
            else:
                result[key]=value
    
    return result


dic1={1:10,2:20}
dic2={3:30,4:40}
dic3={1:20,5:50,6:60}


concatenated_dict =concatenate_dicts(dic1,dic2,dic3)


print("Concatenated Dictionary:",concatenated_dict)



#----------------------------------------------------------
#failed approach
#explored merge function from geeks for geeks
# dic1={1:10, 2:20}
# dic2={3:30, 4:40}
# dic3={1:20,5:50,6:60}
# expected_dict={}
# print(merge(dict1,dict2,dict3))


# another failed approach
# dic1={1:10, 2:20}
# dic2={3:30, 4:40}
# dic3={1:20,5:50,6:60}

# dict4={**dic1,**dic2,**dic3}

# print(dict4)
````

### <u>Question 11:</u>

Write a Python function that takes a list of integers as input and generates a frequency distribution, grouping the integers into ranges of 10. For example, given the input [13, 42, 15, 37, 22, 39, 41, 50], the function should output a dictionary representing the count of integers in each range, such as {'10-19': 2, '20-29': 1, '30-39': 2, '40-49': 2, '50-59': 1}. Implement the function and demonstrate its usage with the provided input. What could be practical use case for this ?

**Input:**
```python
[13,42,15,37,22,39,41,50]
```
**Output:**
```python
{10-19:2,20-29:1,30-39:2,40-49:2,50-59:1}

````python
def frequency_distribution(numbers):
    freq_distribution={}

    for number in numbers:
        lower=(number//10)*10
        upper=lower+9
        range_key=f"{lower}-{upper}"

        if range_key in freq_distribution:
            freq_distribution[range_key]+=1
        else:
            freq_distribution[range_key]=1

    return freq_distribution


numbers=[13,42,15,37,22,39,41,50]


result=frequency_distribution(numbers)


print(result)
````
